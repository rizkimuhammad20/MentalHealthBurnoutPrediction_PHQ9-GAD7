**Encoding**

```python
import pandas as pd
from sklearn.preprocessing import OrdinalEncoder

df = pd.read_csv('Mental Health Burnout Tech.csv')

# Drop unnecessary column
df.drop(columns=['employee_id'], inplace=True)

# =========================
# One-Hot Encoding
# =========================
kolom_nominal = ['gender', 'work_mode', 'therapy_access']

df = pd.get_dummies(
    df,
    columns=kolom_nominal,
    prefix=kolom_nominal,
    drop_first=True
)

# =========================
# Ordinal Encoding
# =========================

ordinal_mapping = {
    'seniority_level': ['Junior', 'Mid', 'Senior', 'Lead', 'Manager', 'Principal'],
    'company_size': ['Startup (1-50)', 'Small (51-200)', 'Mid (201-1000)', 'Large (1001-5000)', 'Enterprise (5000+)'],
    'phq9_category': ['None (0-4)', 'Mild (5-9)', 'Moderate (10-14)', 'Moderately Severe (15-19)', 'Severe (20-27)'], # Fixed categories
    'gad7_category': ['Minimal (0-4)', 'Mild (5-9)', 'Moderate (10-14)', 'Severe (15-21)'],
    'burnout_level': ['Low', 'Moderate', 'High', 'Severe']
}

for col, urutan in ordinal_mapping.items():
    oe = OrdinalEncoder(categories=[urutan])
    df[f'{col}_ord'] = oe.fit_transform(df[[col]])

    # optional drop original column
    # df.drop(columns=[col], inplace=True)

# =========================
# Frequency Encoding
# =========================

freq_map = df['job_role'].value_counts(normalize=True)
df['job_role_freq'] = df['job_role'].map(freq_map)

# =========================
# Target Encoding
# =========================

for col in ['country', 'industry']:
    target_map = df.groupby(col)['burnout_level'].count()
    df[f'{col}_target'] = df[col].map(target_map)

# =========================
# Save file
# =========================

df.to_csv('data_pelanggan_encoded.csv', index=False)

print(df.head())
```
